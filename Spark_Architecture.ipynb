## Introduction
Big Data tools are used to store, process, and analyze large volumes of data. Common tools include:



*   Hadoop: Stores large data using HDFS.
*   Hive: SQL-like querying for data in Hadoop.
*   HBase: NoSQL database on Hadoop.
*   Pig: Scripting platform for processing big data.
*   Sqoop: Transfers data between Hadoop and relational databases.
*   Parquet: Columnar file format optimized for analytics.


## 1. Data Serialization Example
Serialization means converting data into a format that can be stored or transmitted (e.g., JSON, Avro, Parquet). Using JSON and Parquet in Pandas.
Using JSON and Parquet in Pandas.

````python
import pandas as pd
data = {'name': ['Ayesha', 'Ali'], 'age': [14, 30]}
df = pd.DataFrame(data)

# Save to JSON
df.to_json('data.json', orient='records', lines=True)

# Save to Parquet
!pip install pyarrow
df.to_parquet('data.parquet', engine='pyarrow')
````

### Types of File Formats

2.   JSON	JavaScript Object Notation

4.   Parquet	Columnar storage, best for analytics
5.   ORC	Optimized Row Columnar format


## 2. Connect to MySQL and Create Table (Requires MySQL running)


````python
import sqlite3

# Create an in-memory SQLite DB
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# Create table
cursor.execute("CREATE TABLE students (id INTEGER, name TEXT)")

# Insert data
cursor.executemany("INSERT INTO students (id, name) VALUES (?, ?)", [(1, "John"), (2, "Jane")])
conn.commit()

# Query data
cursor.execute("SELECT * FROM students")
rows = cursor.fetchall()
for row in rows:
    print(row)

conn.close()
````

## 6. Sqoop Import to Parquet (Reference Only)
Sqoop transfers data from MySQL to Hadoop in Parquet format.
```bash
sqoop import \
--connect jdbc:mysql://localhost/employees \
--username root \
--password root \
--table emp_data \
--as-parquetfile \
--target-dir /user/hadoop/emp_parquet
```
> Run this in a terminal with Hadoop and Sqoop installed.

## 🔥 Apache Spark Basics
Apache Spark is a powerful open-source engine for large-scale data processing. It works well with big data systems like Hadoop.

### 🏗️ Spark Architecture, Execution, and Concepts
- **Driver**: Main controller of the Spark program
- **Executors**: Run tasks and return results to the driver
- **Cluster Manager**: Allocates resources (e.g., YARN, Mesos)
- **Transformations**: Lazy operations on RDDs
- **Actions**: Triggers execution (e.g., `count()`, `collect()`)

### ⚙️ RDD Operations in Spark
RDD = Resilient Distributed Dataset
- **Transformations**: map(), filter(), flatMap(), union()
- **Actions**: count(), collect(), reduce(), first()

### 🔁 Functional Programming in Spark
Spark uses functions like `map`, `reduce`, and `filter` to operate on data.
These functions are passed as arguments (higher-order functions).

### 🔢 RDD Data Types and Creation
RDDs can hold any type of data: strings, numbers, tuples, custom objects.
To use Spark in Colab, we simulate simple transformations using PySpark.

````python
# 📦 Install PySpark
!pip install pyspark
````

````python
# 🚀 Initialize Spark
from pyspark import SparkContext
sc = SparkContext("local", "SparkBasics")
````

````python
# 📄 Create RDD and Apply Operations
rdd = sc.parallelize([1, 2, 3, 4, 5, 6])
even_rdd = rdd.filter(lambda x: x % 2 == 0)
squared_rdd = even_rdd.map(lambda x: x * x)
result = squared_rdd.collect()
print("✅ Result (Squared Even Numbers):", result)
````
